```python
import os
import pandas as pd
import matplotlib.pyplot as plt
from pptx import Presentation
from pptx.util import Inches
```

```python
# Load data
df = pd.read_csv("data/user_data.csv")
```

```python
# Remove exact duplicate rows
df = df.drop_duplicates()
```

```python
# Group by age and get the average (or total) income
age_income = df.groupby("age")["income"].mean().sort_values(ascending=False)
```

```python
# Ensure directory exists
os.makedirs("charts", exist_ok=True)
```

```python
# Plot the top 10 ages with highest average income
plt.figure(figsize=(10, 6))
age_income.head(10).plot(kind="bar", color='blue')
plt.title("Top 10 Ages with Highest Average Income")
plt.xlabel("Age")
plt.ylabel("Average Income")
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig("charts/top_ages_income.png")
plt.show()
```

```python
# Aggregate spending per category by gender
spending_cols = ["utilities", "entertainment",
                 "school_fees", "shopping", "healthcare"]
gender_spending = df.groupby("gender")[spending_cols].sum()

# Plot
gender_spending.T.plot(kind="bar", figsize=(10, 6))
plt.title("Spending by Gender Across Categories")
plt.xlabel("Spending Category")
plt.ylabel("Total Amount Spent")
plt.legend(title="Gender")
plt.tight_layout()
plt.savefig("charts/spending_by_gender.png")
plt.show()
```

```python
# Create a new PowerPoint presentation
prs = Presentation()

title_slide = prs.slides.add_slide(prs.slide_layouts[0])
title_slide.shapes.title.text = "Data Visualizations"
title_slide.placeholders[1].text = "Presented by Esther Casely Hayford"

# Path to charts folder
charts_folder = "charts"

# Get all PNG files in the charts directory
chart_files = [f for f in os.listdir(charts_folder) if f.endswith(".png")]

# Add each image as a new slide
for chart_file in chart_files:
    slide = prs.slides.add_slide(prs.slide_layouts[5])  # blank layout
    image_path = os.path.join(charts_folder, chart_file)

    # Add the image to the center of the slide
    left = Inches(1)
    top = Inches(1)
    height = Inches(5.5)
    slide.shapes.add_picture(image_path, left, top, height=height)

# Save the presentation
output_path = os.path.join(charts_folder, "visual_report.pptx")
prs.save(output_path)

print(f"PowerPoint created at: {output_path}")
```
